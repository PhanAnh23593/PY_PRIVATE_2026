## Bài 1 · Điền xử lý chuỗi
Chuẩn hóa khoảng trắng trong họ tên và tạo chữ thường. Kết quả: `Nguyen Van An` và `nguyen van an`.

**Đối chiếu:** `Nguyen Van An` và `nguyen van an`.

```python
raw_name = "  Nguyen   Van  An  "
clean_name = " ".join(raw_name.split())
lower_name = clean_name.lower()
print(clean_name)
print(lower_name)
assert clean_name == "Nguyen Van An"
assert lower_name == "nguyen van an"
```

## Bài 2 · Thống kê list
Cho điểm `[8,4,9,5]`. Tính trung bình, cao nhất và số điểm >=5 bằng vòng lặp. Giữ nguyên list gốc.

**Đối chiếu:** Trung bình `6.5`, cao nhất `9`, số điểm đạt `3`.

```python
scores = [8, 4, 9, 5]
total = 0
highest = None
passed_count = 0
for score in scores:
    total += score
    if highest is None or score > highest:
        highest = score
    if score >= 5:
        passed_count += 1
average = total / len(scores) if scores else None
print("Trung binh:", average)
print("Cao nhat:", highest)
print("So diem dat:", passed_count)
assert (average, highest, passed_count) == (6.5, 9, 3)
assert scores == [8, 4, 9, 5]
```

## Bài 3 · Viết mới lọc danh sách
Cho tên có khoảng trắng/trùng sau chuẩn hóa. Tạo unique_names theo thứ tự gặp, bỏ tên rỗng và tên trùng. Chỉ dùng string/list/loop/if; chưa cần set.

**Đối chiếu:** `["Lan", "Minh", "An"]`, giữ thứ tự gặp và bỏ tên rỗng.

```python
raw_names = [" Lan ", "Minh", "Lan", "  ", " An  "]
unique_names = []
for name in raw_names:
    normalized_name = " ".join(name.split())
    if normalized_name and normalized_name not in unique_names:
        unique_names.append(normalized_name)
print(unique_names)
assert unique_names == ["Lan", "Minh", "An"]
assert raw_names == [" Lan ", "Minh", "Lan", "  ", " An  "]
```

## Bài 4 · Đọc/sửa dictionary
Cập nhật điểm Lan từ 7 lên 8.5, thêm city="Thai Nguyen", lấy phone bằng get với mặc định "Chua co".

**Đối chiếu:** score `8.5`, city `Thai Nguyen`, phone `Chua co`.

```python
student = {"id": "SV01", "name": "Lan", "score": 7}
student["score"] = 8.5
student["city"] = "Thai Nguyen"
phone = student.get("phone", "Chua co")
print(student)
print("So dien thoai:", phone)
assert student["score"] == 8.5
assert student["city"] == "Thai Nguyen"
assert phone == "Chua co"
```

## Bài 5 · Set và tuple
Tạo set các môn duy nhất; tạo tuple tên ba môn đã sắp xếp theo chữ cái.

**Đối chiếu:** Set gồm Git, Python, Web; tuple `("Git", "Python", "Web")`.

```python
subjects = ["Python", "Web", "Python", "Git", "Web"]
unique_subjects = set(subjects)
subject_tuple = tuple(sorted(unique_subjects))
print("Cac mon duy nhat:", sorted(unique_subjects))
print(subject_tuple)
assert unique_subjects == {"Git", "Python", "Web"}
assert subject_tuple == ("Git", "Python", "Web")
```

## Bài 6 · Thống kê bản ghi
Tự tạo dict đếm học viên theo group từ list dưới. Không gán đáp án trực tiếp; dùng loop và get.

**Đối chiếu:** `{"Python": 2, "Web": 1}`; list rỗng cho `{}`.

```python
students = [
    {"name": "Lan", "group": "Python"},
    {"name": "Minh", "group": "Web"},
    {"name": "An", "group": "Python"},
]
counts = {}
for student_record in students:
    group = student_record["group"]
    counts[group] = counts.get(group, 0) + 1
print(counts)
assert counts == {"Python": 2, "Web": 1}
```

### Tu kiem tra them
Thu chuoi rong, ten nhieu khoang trang, ten trung va danh sach hoc vien rong.
