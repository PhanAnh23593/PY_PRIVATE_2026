## 1. Ôn nhanh kiến thức

- `== != < <= > >=` so sánh; `and/or/not` kết hợp điều kiện.
- `if/elif/else` chạy nhánh đầu tiên đúng; thứ tự ngưỡng quan trọng.
- Khối code dùng thụt lề nhất quán, thường 4 dấu cách.
- `for` duyệt sequence; `range(1, 6)` lấy 1–5, không lấy 6.
- `while` lặp khi điều kiện còn đúng; luôn nghĩ tới điều kiện dừng.
- `break` thoát vòng lặp gần nhất; `continue` bỏ phần còn lại của lượt hiện tại.
- Biến tích lũy cần khởi tạo ngoài vòng lặp. Theo dõi giá trị từng lượt để tìm lỗi.
- `random.randint(a,b)` lấy số nguyên gồm cả hai đầu; dùng trong trò đoán số mở rộng.

## Bài A · Điền nhánh xếp loại (cơ bản)
Điểm 0–10: >=8 là `Tot`, >=5 là `Dat`, còn lại `Can co gang`. Dùng if/elif/else, không viết trực tiếp đáp án cho điểm mẫu.


```python
score = 8
if score >= 8:
    classification = "Tot"
elif score >= 5:
    classification = "Dat"
else:
    classification = "Can co gang"
print(classification)
```

**Tự kiểm tra:** ô dưới đã được chuẩn bị; hãy chạy sau khi hoàn thành phần TODO.


```python
if classification is None:
    print("Chua hoan thanh Bai A.")
else:
    assert classification == "Tot"
    print("Dat mau diem 8. Tu thu cac moc 0, 5, 7.9, 10.")
```

## Bài B · Tự viết vòng lặp (cơ bản)
Tính tổng các số chẵn từ 1 đến n=10 bằng for. Không dùng công thức hoặc gán trực tiếp 30. Sau đó tự thử n=0 và n=1 → tổng 0.


```python
n = 10
even_total = 0
for number in range(1, n + 1):
    if number % 2 == 0:
        even_total += number
print("Tong cac so chan:", even_total)
```

**Tự kiểm tra:** ô dưới đã được chuẩn bị; hãy chạy sau khi hoàn thành phần TODO.


```python
if even_total is None:
    print("Chua hoan thanh Bai B.")
else:
    assert even_total == 30
    print("Dat du lieu mau; tu thu n=0 va n=1.")
```

## Bài C · while và điều kiện dừng (cơ bản)
Dùng while in đếm ngược 5,4,3,2,1; đếm số lần lặp vào turns. Sau vòng lặp, counter=0 và turns=5.


```python
counter = 5
turns = 0
while counter > 0:
    print(counter)
    counter -= 1
    turns += 1
```

**Tự kiểm tra:** ô dưới đã được chuẩn bị; hãy chạy sau khi hoàn thành phần TODO.


```python
if turns is None:
    print("Chua hoan thanh Bai C.")
else:
    assert counter == 0 and turns == 5
    print("Dat Bai C.")
```

## Bài D · Thử thách
Viết trò đoán số 1–20, giới hạn 5 lượt, thông báo “cao hơn/thấp hơn”, dừng ngay khi đoán đúng. Bản kiểm thử dùng secret=7 để biết kết quả; khi chơi mới dùng random. Giả định đầu vào là số nguyên hợp lệ, chưa cần try/except. Ghi rõ vị trí dùng break và điều kiện dừng.


```python
import random

# guesses la danh sach doan de tu kiem tra; bo qua de nhap khi choi.
def play_guessing_game(secret=None, guesses=None):
    if secret is None:
        secret = random.randint(1, 20)
    won = False
    turns = 0
    # Vong lap dung sau toi da 5 luot hoac khi gap break.
    for turn in range(1, 6):
        if guesses is None:
            guess = int(input("Doan so tu 1 den 20: "))
        else:
            if turn > len(guesses):
                break  # Het du lieu kiem thu.
            guess = guesses[turn - 1]
            print("Luot", turn, "- doan:", guess)
        turns += 1
        if guess == secret:
            print("Chinh xac!")
            won = True
            break  # Doan dung thi dung ngay, khong dung luot con lai.
        elif guess < secret:
            print("Hay doan cao hon.")
        else:
            print("Hay doan thap hon.")
    if not won:
        print("Chua doan dung. Dap an la:", secret)
    return won, turns

assert play_guessing_game(secret=7, guesses=[3, 10, 7, 8]) == (True, 3)
assert play_guessing_game(secret=7, guesses=[1, 2, 3, 4, 5, 7]) == (False, 5)
assert play_guessing_game(secret=7, guesses=[7]) == (True, 1)
# Khi muon choi voi so ngau nhien, bo dau # o dong duoi:
# play_guessing_game()
```

### Tu kiem tra them
Da thu cac moc diem 0, 5, 7.9, 8, 10; tong chan voi n=0, n=1 va n=6.
Tro doan so o tren thu dung o luot 1, dung o luot 3 va het 5 luot.

```python
for test_score, expected in [(0, "Can co gang"), (5, "Dat"), (7.9, "Dat"), (8, "Tot"), (10, "Tot")]:
    if test_score >= 8:
        result = "Tot"
    elif test_score >= 5:
        result = "Dat"
    else:
        result = "Can co gang"
    assert result == expected

for test_n, expected in [(0, 0), (1, 0), (6, 12)]:
    result = 0
    for number in range(1, test_n + 1):
        if number % 2 == 0:
            result += number
    assert result == expected
print("Da qua cac ca kiem tra bo sung.")
```
